## 18. LLM Options

The retrieval pipeline is independent of the final language model, so the project can be completed with a small local model or a larger model when more compute is available.

A practical progression is:

| Level | Example model | Approx. Ollama model size | Use case |
|---|---|---:|---|
| Very small | `llama3.2:1b` | ~1.3 GB | Basic pipeline testing on limited hardware |
| Small | `qwen3.5:2b` | ~2.7–3.1 GB | Lightweight local RAG experiments |
| Recommended default | `qwen3.5:4b` | ~3.3–4.0 GB | Good balance for this project |
| Stronger | `qwen3.5:9b` | ~6.6–7.6 GB | Better answer quality when more memory is available |
| Stronger alternative | `gemma3:12b` | ~8.1 GB | Larger local model for stronger generation |
| Large local option | `qwen3.5:27b` | ~17–20 GB | High-memory workstation or powerful Apple Silicon system |

The sizes above are approximate packaged/quantized Ollama model sizes, not guaranteed RAM or VRAM requirements. Actual memory use depends on quantization, context length, runtime, and hardware.

For this project, `qwen3.5:4b` is a reasonable starting point. A smaller model is completely acceptable when hardware is limited. The goal is to evaluate the RAG workflow, not to use the largest possible model.

References:
- Ollama Qwen3.5 model library: https://ollama.com/library/qwen3.5
- Ollama Llama 3.2 model library: https://ollama.com/library/llama3.2
- Ollama Gemma 3 model library: https://ollama.com/library/gemma3

Ollama exposes a local API at `http://localhost:11434`. Local requests do not require an API key.

A Python client can be installed with:

`pip install ollama`

The final `generate_answer()` function can then call the selected local model.

A CPU-only computer can still run smaller quantized models, but responses may be slower. A GPU or Apple Silicon system generally provides faster local inference. Larger context windows also increase memory use.

### Option B — Hugging Face Transformers

Models can also be loaded directly with `transformers`. This gives more control over the model and generation pipeline, but typically requires more attention to device placement, precision, and available memory.

For this project, Ollama is the simpler local path. Direct Transformers inference is a useful option when more control is needed.
